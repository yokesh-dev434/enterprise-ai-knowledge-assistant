```python
def recursive_split(text, chunk_size, separators):
    chunks=[]
    if len(text) <= chunk_size:
        chunk = text
        chunks.append(chunk)
    else:
        for para in text.split(separators[0]):
            if len(para) <= chunk_size:
                chunk = para
                chunks.append(chunk)
            else:
                for word in para.split(separators[1]):
                    if len(word) <= chunk_size:
                        chunk = word
                        chunks.append(chunk)
                    else:
                        for chara in word:
                            if len(chara) <=chunk_size:
                                chunk = word
                                chunks.append(chunk)
    return chunks


recursive_split(text, chunk_size, separators)
    
```

```python
from qdrant_client import QdrantClient
from qdrant_client.models import Filter, FieldCondition, MatchValue

from app.services.embedding_service import generate_embeddings


# ---------------------------------------
# 1. Connect to Qdrant
# ---------------------------------------

client = QdrantClient(
    host="localhost",
    port=6333
)

collection_name = "knowledge_v1"


# ---------------------------------------
# 2. User query
# ---------------------------------------

query = "What are the standard working hours?"


# ---------------------------------------
# 3. Generate query embedding
# ---------------------------------------

query_embedding = generate_embeddings([query])[0]

print("Query embedding dimension:", len(query_embedding))


# ---------------------------------------
# 4. Create HR metadata filter
# ---------------------------------------

hr_filter = Filter(
    must=[
        FieldCondition(
            key="department",
            match=MatchValue(value="HR")
        )
    ]
)


# ---------------------------------------
# 5. Search Qdrant
# ---------------------------------------

results = client.query_points(
    collection_name=collection_name,
    query=query_embedding.tolist(),
    query_filter=hr_filter,
    limit=5,
    with_payload=True
).points


# ---------------------------------------
# 6. Display results
# ---------------------------------------

print("\nNumber of results:", len(results))

for i, result in enumerate(results, start=1):

    print(f"\n--- Result {i} ---")

    print("Score:", result.score)

    print("Department:",
          result.payload.get("department"))

    print("Document ID:",
          result.payload.get("document_id"))

    print("Source:",
          result.payload.get("source"))

    print("File type:",
          result.payload.get("file_type"))

    print("Text:",
          result.payload.get("text"))
```

```python
from dotenv import load_dotenv
import os
load_dotenv()
# client = genai.Client(
#     api_key=os.getenv("GEMINI_API_KEY")
# )
from typing import Literal
from pydantic import BaseModel, Field

from langchain_google_genai import ChatGoogleGenerativeAI


class DepartmentClassification(BaseModel):
    department: Literal[
        "HR",
        "IT",
        "Finance",
        "Engineering",
        "Client",
        "Product",
        "UNKNOWN"
    ] = Field(
        description="The department most relevant to the user's query."
    )

    confidence: float = Field(
        description="Estimated confidence between 0 and 1."
    )


llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    temperature=0
)

# response =client.models.generate_content(
#         model="gemini-3.1-flash-lite",
#         contents=final_prompt
#     )   
classifier_llm = llm.with_structured_output(
    DepartmentClassification
)


def classify_department(user_query: str) -> DepartmentClassification:
    prompt = f"""
    You are an enterprise query classification system.

    Your task is to classify the user's query into EXACTLY ONE department.

    Allowed departments:

    - HR
    - IT
    - Finance
    - Engineering
    - Client
    - Product
    - UNKNOWN

    ## Classification Rule

    Choose the department whose internal knowledge, documents, policies,
    systems, or expertise would be MOST relevant for answering the user's query.

    Do not classify based only on individual keywords.
    Understand the actual intent of the user's question.

    ## Department Guidelines

    HR:
    Questions about employees, leave, attendance, holidays, payroll-related
    employee policies, performance reviews, recruitment, onboarding,
    benefits, workplace policies, employee conduct, or HR procedures.

    IT:
    Questions about laptops, VPN, passwords, MFA, software installation,
    technical support, IT incidents, access issues, networks, devices,
    or internal IT procedures.

    Finance:
    Questions about expenses, reimbursements, travel expenses,
    procurement, invoices, budgets, payments, payroll finance processes,
    or financial policies.

    Engineering:
    Questions about software development, coding standards, Git,
    APIs, Docker, deployments, databases, architecture, AI development,
    engineering practices, or technical development procedures.

    Client:
    Questions about clients, client requirements, SLAs, support agreements,
    client projects, client-specific documentation, or customer-related
    processes.

    Product:
    Questions about company products, product features, product roadmap,
    product requirements, product behavior, product specifications,
    or product-related processes.

    UNKNOWN:
    Use UNKNOWN when:
    - The query does not clearly belong to any department.
    - The query is unrelated to the enterprise knowledge base.
    - There is insufficient information to determine the department.
    - The query is ambiguous between multiple departments and no department
    is clearly more relevant.

    ## Important Rules

    1. Return EXACTLY ONE department.
    2. Return ONLY the department name.
    3. Do NOT provide explanations.
    4. Do NOT return multiple departments.
    5. Do NOT return JSON.
    6. Do NOT return additional text.
    7. Do NOT invent a department.
    8. Focus on the user's INTENT rather than individual words.

    ## Examples

    User: "How many days of casual leave do employees get?"
    Output:
    HR

    User: "How do I connect to the company VPN?"
    Output:
    IT

    User: "How can I claim my travel expenses?"
    Output:
    Finance

    User: "What is our Git branching strategy?"
    Output:
    Engineering

    User: "What is Client ABC's SLA?"
    Output:
    Client

    User: "What features are planned for the next product release?"
    Output:
    Product

    User: "What is the capital of France?"
    Output:
    UNKNOWN

    User: "Tell me something interesting."
    Output:
    UNKNOWN

    ## User Query

    {user_query}

    ## Output

    Return ONLY one of:

    HR
    IT
    Finance
    Engineering
    Client
    Product
    UNKNOWN
    """
    #     prompt = f"""
    # You are a department classification system for an enterprise AI assistant.

    # Classify the user's query into exactly ONE of these departments:

    # - HR
    # - IT
    # - Finance
    # - Engineering
    # - Client
    # - Product
    # - UNKNOWN

    # Choose the department whose knowledge would be most relevant
    # to answering the user's question.

    # If the query does not belong clearly to any of these departments,
    # return UNKNOWN.

    # User query:
    # {user_query}
    # """

    result = classifier_llm.invoke(prompt)

    return result
```
